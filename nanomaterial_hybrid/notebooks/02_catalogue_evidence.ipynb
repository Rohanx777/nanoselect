# Material catalogue and research evidence
Stage B: all 15 families are assessed for evidence coverage. These are published observations, not trained predictions or preference rankings.

```python
from pathlib import Path
import json
import pandas as pd
ROOT = Path.cwd() if (Path.cwd() / 'data').exists() else Path.cwd().parent
summary = json.loads((ROOT / 'outputs/evidence_audit.json').read_text())
print(summary)
materials = pd.read_csv(ROOT / 'data/materials.csv')
evidence = pd.read_csv(ROOT / 'data/evidence.csv')
coverage = pd.read_csv(ROOT / 'outputs/evidence_coverage.csv')
profiles = pd.read_csv(ROOT / 'data/material_profiles.csv')
sources = pd.read_csv(ROOT / 'data/sources.csv')
```

## Independent study counts
Zero means an evidence gap in this bounded database. It does not mean the material fails. Multiple rows from one paper count once.

```python
coverage.merge(materials[['material_id', 'material_name']], on='material_id').pivot(index=['material_id', 'material_name'], columns='pollutant_id', values='independent_studies')
```
